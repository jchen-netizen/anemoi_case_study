```python
import glob
import os
import re
from collections import defaultdict
from PIL import Image
```

```python
frame_folder = "/Users/jchen/Desktop/ignition_plots"         # folder containing your PNGs
output_folder = "/Users/jchen/Desktop/ignition_plots/gifs"    # where GIFs will be saved
duration = 1000              # ms per frame

os.makedirs(output_folder, exist_ok=True)
```

```python
def group_pngs_by_prefix(frame_folder):
    groups = defaultdict(list)
    all_pngs = glob.glob(os.path.join(frame_folder, "*.png"))

    for path in all_pngs:
        filename = os.path.basename(path)
        name_no_ext = os.path.splitext(filename)[0]
        parts = name_no_ext.split("_")

        if len(parts) < 3:
            print(f"Skipping '{filename}': doesn't match *_*_*.png pattern")
            continue

        prefix = f"{parts[0]}_{parts[1]}"
        groups[prefix].append(path)

    for prefix in groups:
        groups[prefix].sort()

    return groups

groups = group_pngs_by_prefix(frame_folder)

# for prefix, paths in groups.items():
#     print(prefix, "->", len(paths), "frames")
```

```python
def make_gif(image_paths, output_path, duration=1000):
    images = [Image.open(p) for p in image_paths]
    if not images:
        print("No PNG images found in the specified folder.")
        return

    frame_one = images[0]
    frame_one.save(
        output_path,
        format="GIF",
        append_images=images[1:],
        save_all=True,
        duration=duration,
        loop=0
    )
    print(f"GIF successfully saved at: {output_path}")
```

```python
for prefix, paths in groups.items():
    output_path = os.path.join(output_folder, f"{prefix}.gif")
    print(f"\nProcessing group '{prefix}' ({len(paths)} frames)...")
    make_gif(paths, output_path, duration=duration)
```

```python
def make_gif(frame_folder, output_path):
    # Read and sort all PNG images from the directory
    # Sorting ensures your frames play in correct sequence 
    images = [Image.open(image) for image in sorted(glob.glob(f"{frame_folder}/*.png"))]
    
    if not images:
        print("No PNG images found in the specified folder.")
        return

    # Extract the first image to use as the base canvas
    frame_one = images[0]
    
    # Save the sequence as an animated GIF
    frame_one.save(
        output_path, 
        format="GIF", 
        append_images=images[1:], # Append the remaining frames
        save_all=True, 
        duration=500,             # Time per frame in milliseconds (200ms = 0.2 seconds)
        loop=0                    # 0 means loop infinitely, 1 means play once
    )
    print(f"GIF successfully saved at: {output_path}")
```
